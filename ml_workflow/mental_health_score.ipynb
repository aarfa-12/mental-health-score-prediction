### OUTLIER DETECTION

```python
num_col = df.select_dtypes(include='number')
Q1 = num_col.quantile(0.25)
Q3 = num_col.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = (num_col < lower_bound) | (num_col > upper_bound)
print(outliers.sum())
```

```python
skewed_col = ['Study_Hours']
ordinal_col = ['Stress_Level']
categorical_col = ['Gender','Grouped_country','Academic_Level','Most_Used_Platform','Purpose_Of_Use']
numerical_col = ['Age','Avg_Daily_Usage_Hours','Daily_Unlocks','Physical_Activity_Hours','Sleep_Hours_Per_Night']
Features_col = skewed_col + ordinal_col + categorical_col + numerical_col

```

```python
from sklearn.model_selection import train_test_split
y = df['Mental_Health_Score']
X = df[Features_col]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
print(X_train.shape)
print(X_test.shape)
```

### **ML Pipeline**

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer

skew_pipeline = Pipeline(steps=[
    ('log_transform',FunctionTransformer(np.log1p)),
    ('scale',StandardScaler())
    ])

numerical_pipeline = Pipeline(steps=[
    ('scale',StandardScaler())
    ])

ordinal_pipeline = Pipeline(steps=[
    ('ordinal_encoding',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
    ])

categorical_pipeline = Pipeline(steps=[
    ('one_hot_encoding',OneHotEncoder(handle_unknown='ignore'))
    ])


preprocessor = ColumnTransformer(transformers = [
    ('skewed',skew_pipeline,skewed_col),
    ('numerical',numerical_pipeline,numerical_col),
    ('ordinal',ordinal_pipeline,ordinal_col),
    ('categorical',categorical_pipeline,categorical_col)
    
])
```

```python
preprocessor
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

lr_pipeline = Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('regressor',LinearRegression())
])

lr_pipeline.fit(X_train,y_train)
lr_preds = lr_pipeline.predict(X_test)
lr_preds_train  = lr_pipeline.predict(X_train)

r2_testing_lr = r2_score(y_test,lr_preds)
r2_training_lr = r2_score(y_train,lr_preds_train)

mae_testing_lr = mean_absolute_error(y_test,lr_preds)
mae_training_lr = mean_absolute_error(y_train,lr_preds_train)

rmse_testing_lr = np.sqrt(mean_squared_error(y_test,lr_preds))
rmse_training_lr = np.sqrt(mean_squared_error(y_train,lr_preds_train))

print(f'R2 testing accuracy (linear regression): {r2_testing_lr}')
print(f'R2 training accuracy (linear regression): {r2_training_lr}')

print(f'MAE testing accuracy (linear regression):{mae_testing_lr}')
print(f'MAE training accuracy (linear regression): {mae_training_lr}')

print(f'RMSE testing accuracy (linear regression):{rmse_testing_lr}')
print(f'RMSE training accuracy (linear regression): {rmse_training_lr}')
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline = Pipeline(steps = [
    ('preprocessor', preprocessor),
    ('random forest',RandomForestRegressor(random_state=42))
])
rf_pipeline.fit(X_train,y_train)
rf_preds = rf_pipeline.predict(X_test)
rf_preds_train = rf_pipeline.predict(X_train)

r2_testing_rf = r2_score(y_test,rf_preds)
r2_training_rf = r2_score(y_train,rf_preds_train)

mae_testing_rf = mean_absolute_error(y_test,rf_preds)
mae_training_rf = mean_absolute_error(y_train,rf_preds_train)

rmse_testing_rf = np.sqrt(mean_squared_error(y_test,rf_preds))
rmse_training_rf = np.sqrt(mean_squared_error(y_train,rf_preds_train))


print(f'R2 testing accuracy (random forest): {r2_testing_rf}')
print(f'R2 training accuracy (random forest): {r2_training_rf}')

print(f'MAE testing accuracy (random forest):{mae_testing_rf}')
print(f'MAE training accuracy (random forest): {mae_training_rf}')

print(f'RMSE testing accuracy (random forest):{rmse_testing_rf}')
print(f'RMSE training accuracy (random forest): {rmse_training_rf}')
```

### HyperParameter Tuning of Random Forest

```python
from sklearn.model_selection import RandomizedSearchCV

params = {
    'random forest__n_estimators'      : [100,200,300],
    'random forest__max_depth'         : [5,10,15],
    'random forest__min_samples_split' : [2,5,10],
    'random forest__min_samples_leaf'  : [1,2,4]
}

random_search = RandomizedSearchCV(estimator= rf_pipeline,
                   param_distributions = params, 
                   n_iter= 15, 
                   cv= 5, 
                   scoring= 'r2', 
                   random_state=42,
                   n_jobs=-1)

random_search.fit(X_train,y_train)
random_search_preds = random_search.predict(X_test)
random_search_preds_train = random_search.predict(X_train)

r2_testing_rf_tuned = r2_score(y_test,random_search_preds)
r2_training_rf_tuned = r2_score(y_train,random_search_preds_train)

mae_testing_rf_tuned = mean_absolute_error(y_test,random_search_preds)
mae_training_rf_tuned = mean_absolute_error(y_train,random_search_preds_train)

rmse_testing_rf_tuned = np.sqrt(mean_squared_error(y_test,random_search_preds))
rmse_training_rf_tuned = np.sqrt(mean_squared_error(y_train,random_search_preds_train))

print(f'R2 testing accuracy (random forest tuned): {r2_testing_rf_tuned}')
print(f'R2 training accuracy (random forest tuned): {r2_training_rf_tuned}')

print(f'MAE testing accuracy (random forest tuned):{mae_testing_rf_tuned}')
print(f'MAE training accuracy (random forest tuned): {mae_training_rf_tuned}')

print(f'RMSE testing accuracy (random forest tuned):{rmse_testing_rf_tuned}')
print(f'RMSE training accuracy (random forest tuned): {rmse_training_rf_tuned}')

```
